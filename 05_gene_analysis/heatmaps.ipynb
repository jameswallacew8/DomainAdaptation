```python
import pandas as pd
import numpy as np
from pathlib import Path
import gseapy as gsp
from scipy import stats
from scipy.stats import spearmanr
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import seaborn as sns
from itertools import combinations
#  TODO FIXME - requries manual changing of cancer type

# ==========================================
# 1. Load all CSVs into a structured dict
# ==========================================
# Raw attributions: {model_name: {val_tissue: DataFrame}}
all_models = {
    "scGPT ccRCC": file_map_scgpt_alldomain_renal,
    "Geneformer ccRCC": file_map_gf1_alldomain_renal,
    "LightGBM ccRCC": file_map_lgbm_alldomain_renal,
}

# all_models = {
#     "scGPT LUAD": file_map_scgpt_alldomain_lung,
#     "Geneformer LUAD": file_map_gf1_alldomain_lung,
#     "LightGBM LUAD": file_map_lgbm_alldomain_lung,
# }
# all_models = {
#     "scGPT ER+": file_map_scgpt_alldomain_breast,
#     "Geneformer ER+": file_map_gf1_alldomain_breast,
#     "LightGBM ER+": file_map_lgbm_alldomain_breast,
# }
raw_attributions = {}
output_dir = Path("RenalInDomain")
output_dir.mkdir(parents=True, exist_ok=True)
def load_attribution_df(path: str, model_family: str) -> pd.DataFrame:
    """
    Loads and normalises attribution CSVs accounting for model-specific formats.
    Returns a DataFrame with columns: Gene, Score, Rank
    with consistent semantics across model types.
    """
    df = pd.read_csv(path)
    
    if model_family == "LightGBM":
        # Use SHAP as primary score; filter out zero-importance genes
        # (zero SHAP + zero importance = gene did nothing)
        df = df.rename(columns={"Gene": "Gene"})
        df["Score"] = df["SHAP_Importance"]
        df = df[df["LightGBM_Importance"] > 0].copy()  # drop truly unused genes
        df = df.sort_values("Score", ascending=False)
        
    else:  # scGPT, Geneformer
        gene_col = "GeneSymbol" if "GeneSymbol" in df.columns else "Gene"
        # score_col = [c for c in df.columns if c not in [gene_col]][0]
        score_col = "TotalAttributionSum" if "TotalAttributionSum" in df.columns else "SHAP_Importance"
        df = df.rename(columns={gene_col: "Gene", score_col: "Score"})
        # Keep only positive attributions — negatives mean the gene
        # actively pushed against the correct class and are a different
        # biological signal worth analysing separately if needed
        df = df[df["Score"] > 0].copy()
        df = df.sort_values("Score", ascending=False)
    
    df = df[["Gene", "Score"]].dropna().reset_index(drop=True)
    df["Rank"] = df.index + 1
    
    # Min-max normalise scores within each list so cosine similarity
    # is comparable across model types
    score_range = df["Score"].max() - df["Score"].min()
    if score_range > 0:
        df["Score_Norm"] = (df["Score"] - df["Score"].min()) / score_range
    else:
        df["Score_Norm"] = 0.0
    
    return df


# Update the loader in your main loop:
def get_model_family(model_name: str) -> str:
    if "LightGBM" in model_name:
        return "LightGBM"
    elif "scGPT" in model_name:
        return "scGPT"
    elif "Geneformer" in model_name:
        return "Geneformer"
    return "Unknown"

# Replace the loading block in raw_attributions with:
raw_attributions = {}
for model_name, file_map in all_models.items():
    family = get_model_family(model_name)
    raw_attributions[model_name] = {}
    for val_tissue, path in file_map.items():
        df = load_attribution_df(path, family)
        raw_attributions[model_name][val_tissue] = df


# ==========================================
# 2. Stability Analysis — Spearman rank correlations
#    across val datasets within each model
# ==========================================
stability_output = output_dir / "stability"
stability_output.mkdir(exist_ok=True)

stability_records = []

for model_name, val_dict in raw_attributions.items():
    val_tissues = list(val_dict.keys())
    n = len(val_tissues)
    corr_matrix = pd.DataFrame(np.nan, index=val_tissues, columns=val_tissues)

    for t1, t2 in combinations(val_tissues, 2):
        df1 = val_dict[t1].set_index("Gene")["Rank"]
        df2 = val_dict[t2].set_index("Gene")["Rank"]
        common = df1.index.intersection(df2.index)
        if len(common) < 10:
            continue
        rho, _ = spearmanr(df1[common], df2[common])
        corr_matrix.loc[t1, t2] = rho
        corr_matrix.loc[t2, t1] = rho

    np.fill_diagonal(corr_matrix.values, 1.0)
    corr_matrix = corr_matrix.astype(float)
    # Exclude diagonal from mean calculations
    mean_rho = corr_matrix.where(~np.eye(n, dtype=bool)).stack().mean()
    stability_records.append({"Model": model_name, "Mean_Spearman_Rho": mean_rho})

    # Plot heatmap per model
    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm",
                vmin=-1, vmax=1, ax=ax, square=True)
    ax.set_title(f"Inter-val Spearman ρ\n{model_name}", fontsize=10)
    plt.tight_layout()
    safe_name = model_name.replace("\n", "_").replace(" ", "_")
    # fig.savefig(stability_output / f"stability_{safe_name}.png", dpi=150)
    plt.close()

stability_df = pd.DataFrame(stability_records).sort_values("Mean_Spearman_Rho", ascending=False)
stability_df.to_csv(stability_output / "stability_summary.csv", index=False)
print("\nStability summary:")
print(stability_df.to_string(index=False))

# ==========================================
# 4. Enrichment on consensus gene lists (one per model)
# ==========================================
enr_output = output_dir / "enrichment"
enr_output.mkdir(exist_ok=True)

# ==========================================
# 5. Per-val-dataset enrichment + consistency scoring
#    Builds on your original loop — now we also record
#    per-pathway consistency (how many val sets enriched it)
# ==========================================
all_results = []

for model_name, file_map in all_models.items():
    print(f"Processing enrichment: {model_name}...")
    for val_tissue, path in file_map.items():
        df = pd.read_csv(path)
        gene_col = "GeneSymbol" if "GeneSymbol" in df.columns else "Gene"
        top_genes = df.head(50)[gene_col].tolist()

        enr = gsp.enrichr(
            gene_list=top_genes,
            gene_sets=['MSigDB_Hallmark_2020'],
            organism='human',
            cutoff=1.0
        )
        res = enr.results.copy()
        res["Model"] = model_name
        res["Validation_Tissue"] = val_tissue
        all_results.append(res)

df_all = pd.concat(all_results, ignore_index=True)
df_all.to_csv(enr_output / "all_per_val_enrichment.csv", index=False)

# ==========================================
# 6. Pathway consistency heatmap
#    Rows = pathways, Cols = val datasets
#    Cells = -log10(p) so instability is visually obvious
# ==========================================
heatmap_output = output_dir / "heatmaps"
heatmap_output.mkdir(exist_ok=True)

df_all["-log10p"] = -np.log10(df_all["P-value"].clip(lower=1e-10))

# Summarise using Fisher's combined p (Stouffer could swap in here)
def fishers_combined(pvals):
    pvals = np.clip(pvals, 1e-10, 1.0)
    stat = -2 * np.sum(np.log(pvals))
    combined_p = stats.chi2.sf(stat, df=2 * len(pvals))
    return combined_p

# Count consistency: fraction of val datasets where p < 0.05
def consistency_score(pvals):
    return np.mean(np.array(pvals) < 0.05)

for model_name in df_all["Model"].unique():
    model_df = df_all[df_all["Model"] == model_name]
    
    pivot = model_df.pivot_table(
        index="Term", columns="Validation_Tissue",
        values="-log10p", aggfunc="mean"
    ).fillna(0)
    
    # Filter to pathways enriched (p<0.05) in at least 1 val dataset
    pivot_raw_p = model_df.pivot_table(
        index="Term", columns="Validation_Tissue",
        values="P-value", aggfunc="mean"
    ).fillna(1.0)
    keep = (pivot_raw_p < 0.05).any(axis=1)
    pivot = pivot[keep]
    pivot_raw_p = pivot_raw_p[keep]
    
    if pivot.empty:
        continue
    
    # Sort pathways by consistency (most consistent at top)
    pivot["_consistency"] = pivot_raw_p.apply(lambda row: consistency_score(row.values), axis=1)
    pivot = pivot.sort_values("_consistency", ascending=False).drop(columns="_consistency")
    pivot_raw_p = pivot_raw_p.loc[pivot.index]
    
    fig, ax = plt.subplots(figsize=(max(8, len(pivot.columns) * 1.2), max(6, len(pivot) * 0.4)))
    sns.heatmap(
        pivot, cmap="YlOrRd", ax=ax,
        linewidths=0.3, linecolor="grey",
        cbar_kws={"label": "-log10(p-value)"}
    )
    # Annotate Fisher's p on the right
   
    safe_name = model_name.replace("\n", "_").replace(" ", "_")
    ax.set_title(f"Pathway Enrichment Consistency\n{model_name}", fontsize=11)
    plt.tight_layout()
    # fig.savefig(heatmap_output / f"pathway_heatmap_{safe_name}.png", dpi=300, bbox_inches="tight")
    plt.close()
    print(f"  Saved heatmap: {model_name}")
```

```python
from matplotlib.gridspec import GridSpec

#  TODO FIXME - requries manual changing of cancer type
# ==========================================
# Combined heatmaps — one per train condition
# 3 panels side by side (scGPT | Geneformer | LightGBM)
# Shared pathway axis = union of top pathways per model
# ==========================================

TOP_PATHWAYS_PER_MODEL = 8  # top N per model before taking union → max 30 rows
MODELS = ["Geneformer", "scGPT", "LightGBM"]
TRAIN_CONDITIONS = ['ccRCC']

VAL_TISSUE_ORDER = {
    "ccRCC": ["Cambridge", "US", "Lithuania", "China", "Metastasis", "Chromophobe"],
    "LUAD":  ["LUAD_GSE148071", "LUAD_GSE117570", "LUAD_GSE127465", "LUAD_LBM", "NSCLC_LM", "LUSC_GSE117570", "LUSC_GSE127465", 'LUSC_GSE148071', 'SCLC_GSE150766'],
    "Breast":            ["Breast", "ccRCC", "Chromophobe", "Colorectal", "Lung", "Metastasis"],
    "ER+": ["GSE176078 ER+",
            "GSE161529 ER+",
            "PRJNA1140267 ER+",
            "EMTAB8107 Luminal",
            "GSE161529 HER2+",
            "GSE176078 HER2+",
            "GSE161529 TNBC BRCA1",
            "GSE161529 TNBC",
            "GSE176078 TNBC"]
}

VAL_TISSUE_SUBTYPES = {
    # Lung
    "LUAD_GSE148071":  "LUAD",
    "LUAD_GSE117570":  "LUAD",
    "LUAD_GSE127465":  "LUAD",
    "LUAD_LBM":        "LUAD",
    "NSCLC_LM":        "LUAD",
    "LUSC_GSE117570":  "LUSC",
    "LUSC_GSE127465":  "LUSC",
    "LUSC_GSE148071":  "LUSC",
    "SCLC_GSE150766":  "SCLC",
    # Kidney
    "Cambridge":           "ccRCC",
    "US":           "ccRCC",
    "Lithuania":       "ccRCC",
    "China":           "ccRCC",
    "Metastasis":      "Metastatic",
    "Chromophobe":     "ChRCC",
    # Breast
    "GSE176078 ER+": "ER+",
    "GSE161529 ER+": "ER+",
    "PRJNA1140267 ER+": "ER+",
    "EMTAB8107 Luminal": "ER+",
    "GSE161529 HER2+": "HER2+",
    "GSE176078 HER2+": "HER2+",
    "GSE161529 TNBC BRCA1": "TNBC",
    "GSE161529 TNBC": "TNBC",
    "GSE176078 TNBC": "TNBC",    
}

SUBTYPE_COLOURS = {
    "LUAD":        "#72aaf1",
    "LUSC":        "#e68e36",
    "SCLC":        "#62e049",

    "ccRCC":       "#72aaf1",
    "Metastatic":  "#e68e36",
    "ChRCC": "#62e049",

    "ER+":        "#72aaf1",
    "HER2+":        "#e68e36",
    "TNBC":        "#62e049",
}

# ---- Replace the all_val_cols derivation (step 4) with: ----
# ==========================================
# Performance correlation with pathway enrichment
# Added as an extra column to the combined heatmaps
# ==========================================

# Structure your performance scores as a nested dict:
# {model_key: {val_tissue: auroc}}
# e.g. performance_scores["scGPT\n(ccRCC)"]["Lung"] = 0.61
# Already defined earlier — just make sure all 9 model keys are filled in

from scipy.stats import spearmanr


# Pre-compute -log10p and Fisher's p in df_all if not already done
df_all["-log10p"] = -np.log10(df_all["P-value"].clip(lower=1e-10))

def fishers_combined(pvals):
    pvals = np.clip(pvals, 1e-10, 1.0)
    stat = -2 * np.sum(np.log(pvals))
    return stats.chi2.sf(stat, df=2 * len(pvals))

combined_output = output_dir / "combined_heatmaps"
combined_output.mkdir(exist_ok=True)

for train_cond in TRAIN_CONDITIONS:

    # ---- 1. Identify the three model keys for this train condition ----
    model_keys = {
        family: f"{family} {train_cond}"
        for family in MODELS
    }
    # Verify they exist in df_all
    model_keys = {
        fam: key for fam, key in model_keys.items()
        if key in df_all["Model"].unique()
    }
    if not model_keys:
        print(f"No models found for train condition: {train_cond}")
        continue

    # ---- 2. Build per-model pivot tables (pathway × val_tissue, values = -log10p) ----
    pivots = {}
    fisher_ranks = {}  # pathway → Fisher p, per model

    for family, model_key in model_keys.items():
        model_df = df_all[df_all["Model"] == model_key]

        pivot = model_df.pivot_table(
            index="Term", columns="Validation_Tissue",
            values="-log10p", aggfunc="mean"
        ).fillna(0)

        pivot_p = model_df.pivot_table(
            index="Term", columns="Validation_Tissue",
            values="P-value", aggfunc="mean"
        ).fillna(1.0)

        # Fisher's combined p per pathway
        pivot_p["Fisher_p"] = pivot_p.apply(
            lambda row: fishers_combined(row.values), axis=1
        )

        pivots[family] = pivot
        fisher_ranks[family] = pivot_p["Fisher_p"].sort_values()

    # ---- 3. Select shared pathway axis ----
    # Take top N pathways per model by Fisher's p, then union
    selected_pathways = set()
    for family in model_keys:
        top = fisher_ranks[family].head(TOP_PATHWAYS_PER_MODEL).index.tolist()
        selected_pathways.update(top)
    selected_pathways = sorted(selected_pathways)  # alphabetical for consistency

    # Re-order pathways by mean Fisher rank across models (best combined evidence at top)
    pathway_mean_fisher = {}
    for pathway in selected_pathways:
        fps = []
        for family in model_keys:
            fp = fisher_ranks[family].get(pathway, 1.0)  # 1.0 if not in model's results
            fps.append(fp)
        pathway_mean_fisher[pathway] = np.mean(fps)

    selected_pathways = sorted(selected_pathways, key=lambda p: pathway_mean_fisher[p])

    # ---- 4. Build final padded pivots aligned to shared pathway axis ----
    # Determine shared val tissue columns (should be same 6 for all models)
  
    available_cols = set(
        col for fam in model_keys
        for col in pivots[fam].columns
    )

    # Apply custom order if defined, append any unlisted cols at end
    preferred_order = VAL_TISSUE_ORDER.get(train_cond, [])
    all_val_cols = (
        [c for c in preferred_order if c in available_cols] +
        sorted([c for c in available_cols if c not in preferred_order])
    )

    print(f"Val tissue order for {train_cond}: {all_val_cols}")

    aligned_pivots = {}
    for family in model_keys:
        p = pivots[family].reindex(index=selected_pathways, columns=all_val_cols).fillna(0)
        aligned_pivots[family] = p
    

    # ---- 5. Compute shared colour scale ----
    vmax = max(p.values.max() for p in aligned_pivots.values())
    vmin = 0

    # ---- 6. Plot ----
    n_models = len(model_keys)
    n_pathways = len(selected_pathways)
    n_val = len(all_val_cols)

    # Width: each model panel gets (n_val * 0.7) inches + 0.3 gap
    # Height: pathways * 0.35 inches
    panel_width = n_val * 0.8
    fig_width = panel_width * n_models + 2.5  # extra for cbar + labels
    fig_height = max(6, n_pathways * 0.38 + 1.5)

    fig = plt.figure(figsize=(fig_width, fig_height))
    # Two rows: thin annotation strip on top, heatmap below
    ANNOT_HEIGHT = 0.18   # inches — controls strip thickness
    gs = GridSpec(
        2, n_models + 1,
        figure=fig,
        height_ratios=[ANNOT_HEIGHT, fig_height - ANNOT_HEIGHT],
        width_ratios=[panel_width] * n_models + [0.3],
        wspace=0.05,
        hspace=0.01
    )

    annot_axes  = [fig.add_subplot(gs[0, i]) for i in range(n_models)]
    heatmap_axes = [fig.add_subplot(gs[1, i]) for i in range(n_models)]
    cbar_ax     = fig.add_subplot(gs[1, n_models])


    # Clean pathway names (MSigDB names tend to be long with underscores)
    def clean_pathway(name):
        return name.replace("HALLMARK_", "").replace("_", " ").title()

    y_labels = [clean_pathway(p) for p in selected_pathways]
    strip_colours = [
            SUBTYPE_COLOURS.get(VAL_TISSUE_SUBTYPES.get(col, ""), "#dddddd")
            for col in all_val_cols
        ]
    
    for i, (family, ax) in enumerate(zip(model_keys.keys(), heatmap_axes)):
        pivot = aligned_pivots[family]

        im = ax.imshow(
            pivot.values,
            aspect="auto",
            cmap="YlOrRd",
            vmin=vmin,
            vmax=vmax,
            interpolation="none"
        )

        # X axis: val tissue labels
        ax.set_xticks(range(n_val))
        ax.set_xticklabels(
             all_val_cols,
            rotation=45, ha="right", fontsize=8
        )
        # Y axis: only on leftmost panel
        if i == 0:
            ax.set_yticks(range(n_pathways))
            ax.set_yticklabels(y_labels, fontsize=8)
        else:
            ax.set_yticks([])

        # Significance markers — asterisk where any val p < 0.05
        model_key = model_keys[family]
        model_df = df_all[df_all["Model"] == model_key]
        pivot_p = model_df.pivot_table(
            index="Term", columns="Validation_Tissue",
            values="P-value", aggfunc="mean"
        ).reindex(index=selected_pathways, columns=all_val_cols).fillna(1.0)

        for row_idx, pathway in enumerate(selected_pathways):
            for col_idx, val_tissue in enumerate(all_val_cols):
                p_val = pivot_p.loc[pathway, val_tissue] if pathway in pivot_p.index else 1.0
                # if fp < 0.001: 
                #     ax.text(col_idx, row_idx, "++", ha="center", va="center",
                #             fontsize=9, color="black", fontweight="bold")
                # elif p_val < 0.01:
                #     ax.text(col_idx, row_idx, "++", ha="center", va="center",
                #             fontsize=9, color="black", fontweight="bold")
                if p_val < 0.05:
                    ax.text(col_idx, row_idx, "+", ha="center", va="center",
                            fontsize=9, color="black", fontweight="bold")
                


        # Grid lines
        ax.set_xticks(np.arange(-0.5, n_val, 1), minor=True)
        ax.set_yticks(np.arange(-0.5, n_pathways, 1), minor=True)
        ax.grid(which="minor", color="white", linewidth=0.5)
        ax.tick_params(which="minor", bottom=False, left=False)

        aax = annot_axes[i]
        for xi, colour in enumerate(strip_colours):
            aax.add_patch(plt.Rectangle(
                (xi - 0.5, 0), 1, 1,
                color=colour, transform=aax.transData, clip_on=True
            ))

        # Subtype label centred over each group of same-subtype columns
        seen_subtypes = {}
        for xi, col in enumerate(all_val_cols):
            st = VAL_TISSUE_SUBTYPES.get(col, "")
            seen_subtypes.setdefault(st, []).append(xi)

        for st, positions in seen_subtypes.items():
            if not st:
                continue
            mid = np.mean(positions)
            aax.text(mid, 0.5, st,
                     ha="center", va="center", fontsize=7,
                     fontweight="bold",
                     color="black",
                     transform=aax.transData)

        aax.set_xlim(-0.5, n_val - 0.5)
        aax.set_ylim(0, 1)
        aax.axis("off")
        aax.set_title(family, fontsize=10, fontweight="bold", pad=4)

    # Shared colorbar
    plt.colorbar(im, cax=cbar_ax, label="-log10(p-value)")

    fig.suptitle(
        f"Pathway Enrichment — Train: {train_cond}",
        fontsize=12, fontweight="bold"
    )

    plt.tight_layout()
    fig.savefig(
        combined_output / f"combined_heatmap_{train_cond}.svg", bbox_inches="tight"
    )
    plt.close()
    print(f"Saved combined heatmap for train condition: {train_cond}")
```
