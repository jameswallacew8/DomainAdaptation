```python
import pandas as pd
import numpy as np
from pathlib import Path
import gseapy as gsp
from scipy import stats
from scipy.stats import spearmanr
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import seaborn as sns
from itertools import combinations

# ==========================================
# 1. Load all CSVs into a structured dict
# ==========================================
# Raw attributions: {model_name: {val_tissue: DataFrame}}

# all_models = {
#     "scGPT ccRCC": file_map_scgpt_ccrcc_in_domain,
#     "Geneformer ccRCC": file_map_gf1_ccrcc_in_domain,
#     "LightGBM ccRCC": file_map_lgbm_ccrcc_in_domain,

#      "scGPT LUAD": file_map_scgpt_indomain_luad_train,
#     "Geneformer LUAD": file_map_gf1_indomain_luad_train,
#     "LightGBM LUAD": file_map_lgbm_indomain_luad_train,

#     "scGPT ER+": file_map_scgpt_indomain_breast,
#     "Geneformer ER+": file_map_gf1_indomain_breast,
#     "LightGBM ER+": file_map_lgbm_indomain_breast,
# }

all_models = {
    "scGPT ccRCC": file_map_scgpt_alldomain_renal,
    "Geneformer ccRCC": file_map_gf1_alldomain_renal,
    "LightGBM ccRCC": file_map_lgbm_alldomain_renal,

    "scGPT LUAD": file_map_scgpt_alldomain_lung,
    "Geneformer LUAD": file_map_gf1_alldomain_lung,
    "LightGBM LUAD": file_map_lgbm_alldomain_lung,

    "scGPT ER+": file_map_scgpt_alldomain_breast,
    "Geneformer ER+": file_map_gf1_alldomain_breast,
    "LightGBM ER+": file_map_lgbm_alldomain_breast,
}

VAL_TISSUE_SUBTYPES = {
    # Lung
    "LUAD_GSE148071":  "LUAD",
    "LUAD_GSE117570":  "LUAD",
    "LUAD_GSE127465":  "LUAD",
    "LUAD_LBM":        "LUAD",
    "NSCLC_LM":        "LUAD",
    "LUSC_GSE117570":  "LUSC",
    "LUSC_GSE127465":  "LUSC",
    "LUSC_GSE148071":  "LUSC",
    "SCLC_GSE150766":  "SCLC",
    # Kidney
    "Cambridge":           "ccRCC",
    "US":           "ccRCC",
    "Lithuania":       "ccRCC",
    "China":           "ccRCC",
    "Metastasis":      "Metastatic",
    "Chromophobe":     "ChRCC",
    # Breast
    "GSE176078 ER+": "ER+",
    "GSE161529 ER+": "ER+",
    "PRJNA1140267 ER+": "ER+",
    "EMTAB8107 Luminal": "ER+",
    "GSE161529 HER2+": "HER2+",
    "GSE176078 HER2+": "HER2+",
    "GSE161529 TNBC BRCA1": "TNBC",
    "GSE161529 TNBC": "TNBC",
    "GSE176078 TNBC": "TNBC",    
}
SUBTYPE_COLOURS = {
    "ER+":  "#72aaf1",
    "HER2+": "#e68e36",
    "TNBC":  "#62e049",
}

SUBTYPE_ORDER = ["ccRCC", "LUAD", "ER+"]

output_dir = Path("AllStability")
output_dir.mkdir(parents=True, exist_ok=True)
def load_attribution_df(path: str, model_family: str) -> pd.DataFrame:
    """
    Loads and normalises attribution CSVs accounting for model-specific formats.
    Returns a DataFrame with columns: Gene, Score, Rank
    with consistent semantics across model types.
    """
    df = pd.read_csv(path)
    
    if model_family == "LightGBM":
        # Use SHAP as primary score; filter out zero-importance genes
        # (zero SHAP + zero importance = gene did nothing)
        df = df.rename(columns={"Gene": "Gene"})
        df["Score"] = df["SHAP_Importance"]
        df = df[df["LightGBM_Importance"] > 0].copy()  # drop truly unused genes
        df = df.sort_values("Score", ascending=False)
        
    else:  # scGPT, Geneformer
        gene_col = "GeneSymbol" if "GeneSymbol" in df.columns else "Gene"
        # score_col = [c for c in df.columns if c not in [gene_col]][0]
        score_col = "TotalAttributionSum" if "TotalAttributionSum" in df.columns else "SHAP_Importance"
        df = df.rename(columns={gene_col: "Gene", score_col: "Score"})
        # Keep only positive attributions — negatives mean the gene
        # actively pushed against the correct class and are a different
        # biological signal worth analysing separately if needed
        df = df[df["Score"] > 0].copy()
        df = df.sort_values("Score", ascending=False)
    
    df = df[["Gene", "Score"]].dropna().reset_index(drop=True)
    df["Rank"] = df.index + 1
    
    # Min-max normalise scores within each list so cosine similarity
    # is comparable across model types
    score_range = df["Score"].max() - df["Score"].min()
    if score_range > 0:
        df["Score_Norm"] = (df["Score"] - df["Score"].min()) / score_range
    else:
        df["Score_Norm"] = 0.0
    
    return df


# Update the loader in your main loop:
def get_model_family(model_name: str) -> str:
    if "LightGBM" in model_name:
        return "LightGBM"
    elif "scGPT" in model_name:
        return "scGPT"
    elif "Geneformer" in model_name:
        return "Geneformer"
    return "Unknown"

# Replace the loading block in raw_attributions with:
raw_attributions = {}
for model_name, file_map in all_models.items():
    family = get_model_family(model_name)
    raw_attributions[model_name] = {}
    for val_tissue, path in file_map.items():
        df = load_attribution_df(path, family)
        raw_attributions[model_name][val_tissue] = df


# ==========================================
# 2. Stability Analysis — Spearman rank correlations
#    across val datasets within each model
# ==========================================
stability_output = output_dir / "stability"
stability_output.mkdir(exist_ok=True)

stability_records = []

def order_by_subtype(val_tissues, subtype_map, subtype_order):
    """Return val_tissues sorted by subtype block order."""
    def sort_key(t):
        st = subtype_map.get(t, "zzz")  # unknown subtypes go last
        block = subtype_order.index(st) if st in subtype_order else len(subtype_order)
        return (block, t)
    return sorted(val_tissues, key=sort_key)


for model_name, val_dict in raw_attributions.items():
    val_tissues = list(val_dict.keys())
    n           = len(val_tissues)
    corr_matrix = pd.DataFrame(np.nan, index=val_tissues, columns=val_tissues)

    for t1, t2 in combinations(val_tissues, 2):
        df1    = val_dict[t1].set_index("Gene")["Rank"]
        df2    = val_dict[t2].set_index("Gene")["Rank"]
        common = df1.index.intersection(df2.index)
        if len(common) < 10:
            continue
        rho, _                  = spearmanr(df1[common], df2[common])
        corr_matrix.loc[t1, t2] = rho
        corr_matrix.loc[t2, t1] = rho

    np.fill_diagonal(corr_matrix.values, 1.0)
    corr_matrix = corr_matrix.astype(float)

    mean_rho = corr_matrix.where(~np.eye(n, dtype=bool)).stack().mean()
    stability_records.append({"Model": model_name, "Mean_Spearman_Rho": mean_rho})

    # --- Reorder rows/cols by subtype block ---
    ordered = order_by_subtype(val_tissues, VAL_TISSUE_SUBTYPES, SUBTYPE_ORDER)
    corr_matrix = corr_matrix.reindex(index=ordered, columns=ordered)

    # --- Lower triangle only (mask upper + diagonal) ---
    mask = np.triu(np.ones(corr_matrix.shape, dtype=bool), k=0)  # mask diagonal and above

    fig, ax = plt.subplots(figsize=(max(6, n * 0.7), max(5, n * 0.65)))

    sns.heatmap(
        corr_matrix,
        mask=mask,
        annot=True, fmt=".2f",
        cmap="coolwarm", vmin=-1, vmax=1,
        ax=ax, square=True,
        linewidths=0.5, linecolor="white",
        cbar_kws={"shrink": 0.6, "label": "Spearman ρ"},
        annot_kws={"size": 8}
    )

    # --- Subtype block boundary lines ---
    subtypes_ordered = [VAL_TISSUE_SUBTYPES.get(t, "") for t in ordered]
    boundaries = []
    for idx in range(1, len(subtypes_ordered)):
        if subtypes_ordered[idx] != subtypes_ordered[idx - 1]:
            boundaries.append(idx)

    for b in boundaries:
        ax.axhline(b, color="black", linewidth=2.0, zorder=5)
        ax.axvline(b, color="black", linewidth=2.0, zorder=5)

    # --- Subtype colour strip on left axis ---
    for idx, tissue in enumerate(ordered):
        st     = VAL_TISSUE_SUBTYPES.get(tissue, "")
        colour = SUBTYPE_COLOURS.get(st, "#dddddd")
        ax.add_patch(plt.Rectangle(
            (-0.35, idx), 0.25, 1,
            color=colour,
            transform=ax.transData,
            clip_on=False
        ))

    # --- Subtype block label on diagonal (upper area, now empty) ---
    seen = {}
    for idx, tissue in enumerate(ordered):
        st = VAL_TISSUE_SUBTYPES.get(tissue, "")
        seen.setdefault(st, []).append(idx)

    for st, positions in seen.items():
        mid    = np.mean(positions)
        colour = SUBTYPE_COLOURS.get(st, "grey")
        ax.text(
            mid, mid, st,
            ha="center", va="center",
            fontsize=9, fontweight="bold",
            color=colour, alpha=0.6,
            transform=ax.transData
        )

    ax.set_title(
        f"Inter-val Spearman ρ — {model_name}\nMean ρ = {mean_rho:.3f}",
        fontsize=10, pad=10
    )
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="right", fontsize=8)
    ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=8)

    plt.tight_layout()
    safe_name = model_name.replace("\n", "_").replace(" ", "_")
    # fig.savefig(stability_output / f"stability_{safe_name}.png", dpi=150, bbox_inches="tight")
    plt.close()

# ==========================================
# 3. Aggregate Barplot of Mean Stability
# ==========================================

# 1. Convert the collected stability records into a DataFrame
df_stability = pd.DataFrame(stability_records)

model_color_map = {
    "Geneformer": "#2ECC71",
    "scGPT":      "#3498DB",
    "LightGBM":   "#E74C3C",
}
# 2. Split the "Model" column (e.g., "scGPT ER+") into "Model_Family" and "Dataset"
# n=1 ensures it only splits on the first space, keeping dataset names intact
df_stability[['Model_Family', 'Dataset']] = df_stability['Model'].str.split(' ', n=1, expand=True)

# 3. Create the grouped barplot
fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=df_stability, 
    x='Dataset', 
    y='Mean_Spearman_Rho', 
    hue='Model_Family',
    palette=model_color_map, # Apply the custom colors here
    ax=ax
)

# 4. Formatting
# 4. Add the mean values above each bar
# ax.containers holds the bar coordinates drawn by seaborn
for container in ax.containers:
    ax.bar_label(
        container, 
        fmt='%.3f',   # Format to 3 decimal places
        padding=3,    # Small gap between the top of the bar and the text
        fontsize=9, 
        color='black',
        fontweight='heavy' 
    )
ax.set_title('Aggregate Mean Attribution Stability (Spearman ρ)', fontsize=14, pad=15)
ax.set_ylabel('Mean Spearman ρ', fontsize=12)
ax.set_xlabel('Train Dataset', fontsize=12)

y_max = df_stability['Mean_Spearman_Rho'].max()
ax.set_ylim(0.5, 1.0)

# Move the legend outside the plot so it doesn't overlap the bars
plt.legend( loc='lower left')

plt.tight_layout()

# 5. Save the aggregate plot
barplot_path = stability_output / "Figure3_aggregated_mean_stability_barplot.svg"
plt.savefig(barplot_path, bbox_inches='tight')
plt.close()

print(f"Aggregate barplot saved to: {barplot_path}")
```

```python
# TODO FIXME bad run order in cells above
# ==========================================
# 4. Venn Diagrams of Consensus Genes (Top 250 by Rank)
# ==========================================
import pandas as pd
from matplotlib_venn import venn3

def compute_consensus(gene_lists: list, min_datasets: int = 3) -> pd.DataFrame:
    from collections import Counter
    freq        = Counter(g for lst in gene_lists for g in lst)
    rank_sums   = {}
    rank_counts = {}
    for lst in gene_lists:
        for rank, gene in enumerate(lst, 1):
            rank_sums[gene]   = rank_sums.get(gene, 0) + rank
            rank_counts[gene] = rank_counts.get(gene, 0) + 1

    records = [
        {"Gene": g, "N_Datasets": c,
         "Mean_Rank":  rank_sums[g] / rank_counts[g],
         "Freq_Frac":  c / len(gene_lists)}
        for g, c in freq.items() if c >= min_datasets
    ]
    if not records:
        return pd.DataFrame(columns=["Gene", "N_Datasets", "Mean_Rank", "Freq_Frac"])
    return pd.DataFrame(records).sort_values(
        ["N_Datasets", "Mean_Rank"], ascending=[False, True]
    ).reset_index(drop=True)


# 1. Group the genes by Dataset -> Model Family
dataset_consensus_genes = {}

for model_key, val_dict in raw_attributions.items():
    parts = model_key.split(' ', 1)
    if len(parts) != 2:
        continue
        
    model_family, dataset = parts[0], parts[1]
    
    if dataset not in dataset_consensus_genes:
        dataset_consensus_genes[dataset] = {}
        
    # Extract the ordered gene lists for each validation set
    gene_lists = []
    for val_tissue, df in val_dict.items():
        # df is already sorted by importance/score from load_attribution_df
        gene_lists.append(df['Gene'].tolist())
        
    if gene_lists:
        # Pass min_datasets=1 as a safety net. The function still sorts N_Datasets 
        # descending, so universally present genes will always rise to the top.
        consensus_df = compute_consensus(gene_lists, min_datasets=1)
        
        # Take the top 250 genes based on the rank consensus
        top_250_genes = set(consensus_df.head(250)['Gene'])
        
        dataset_consensus_genes[dataset][model_family] = top_250_genes

# 2. Plot the Venn diagrams (one for each dataset)
num_datasets = len(dataset_consensus_genes)
fig, axes = plt.subplots(1, num_datasets, figsize=(4 * num_datasets, 5))

# Force axes into a list if there's only one dataset active
if num_datasets == 1:
    axes = [axes]

target_models = ["scGPT", "Geneformer", "LightGBM"]
venn_colors = (
    model_color_map["scGPT"], 
    model_color_map["Geneformer"], 
    model_color_map["LightGBM"]
)

for ax, (dataset, models_dict) in zip(axes, dataset_consensus_genes.items()):
    sets = [models_dict.get(m, set()) for m in target_models]
    
    v = venn3(
        subsets=sets, 
        set_labels=target_models, 
        set_colors=venn_colors,
        alpha=0.6,
        ax=ax
    )
    
    ax.set_title(f"{dataset} Domain", fontsize=14, fontweight='bold', pad=15)

plt.tight_layout()

# 3. Save the plot
venn_path = output_dir / "Figure4_top250_consensus_genes_venn.svg"
plt.savefig(venn_path, bbox_inches="tight")
plt.close()

print(f"Venn diagrams saved to: {venn_path}")

# ==========================================
# 4. Extract and Save Intersection Genes
# ==========================================
intersection_dir = output_dir / "consensus_intersections"
intersection_dir.mkdir(exist_ok=True)

for dataset, models_dict in dataset_consensus_genes.items():
    # Safely extract the sets for each specific model
    set_scgpt = models_dict.get("scGPT", set())
    set_gf = models_dict.get("Geneformer", set())
    set_lgbm = models_dict.get("LightGBM", set())
    
    # 1. Calculate Core Intersection (genes in ALL three)
    core_intersection = set_scgpt & set_gf & set_lgbm
    
    # 2. Calculate FM-Only Intersection (genes in scGPT AND Geneformer, but NOT LightGBM)
    fm_only_intersection = (set_scgpt & set_gf) - set_lgbm
    
    # --- Convert to sorted lists ---
    core_list = sorted(list(core_intersection))
    fm_only_list = sorted(list(fm_only_intersection))
    
    # --- Save Core Intersection ---
    if core_list:
        df_core = pd.DataFrame({"Consensus_Gene": core_list})
        core_csv_path = intersection_dir / f"{dataset}_core_intersection_genes.csv"
        df_core.to_csv(core_csv_path, index=False)
        
    # --- Save FM-Only Intersection ---
    if fm_only_list:
        df_fm = pd.DataFrame({"Consensus_Gene": fm_only_list})
        fm_csv_path = intersection_dir / f"{dataset}_scGPT_Geneformer_only.csv"
        df_fm.to_csv(fm_csv_path, index=False)
        
    print(f"[{dataset}] Extracted {len(core_list)} core genes and {len(fm_only_list)} FM-only genes.")
```

```python
# ==========================================
# 5. Subtype Adaptation Analysis
# Rank-Shift Scatter Plots (Gene-level)
# ==========================================
#   FIXME suboptimal run order, must run first cell, ignore second path cell and venn cell above, then this (to get all attr csvs for train condition)

import matplotlib.patches as mpatches
import textwrap
from adjustText import adjust_text
 

# baseline_subtype = "ER+"
# rare_subtypes = ["HER2+", "TNBC"]
baseline_subtype = "ccRCC"
rare_subtypes = ["Metastatic", "ChRCC"]
# baseline_subtype = "LUAD"
# rare_subtypes = ["LUSC", "SCLC"]


# 1. Calculate aggregated mean scores for each Model + Subtype combination
# Structure: {model_family: {subtype: df_mean_scores}}
subtype_mean_scores = {}
# 1. Calculate aggregated mean scores for each Model + Subtype combination
# Structure: {model_family: {subtype: df_mean_scores}}
subtype_mean_scores = {}

for model_key, val_dict in raw_attributions.items():
    model_family = model_key.split(' ', 1)[0]
    if model_family not in subtype_mean_scores:
        subtype_mean_scores[model_family] = {}
        
    # Group validation DataFrames by their biological subtype
    subtype_dfs = {}
    for val_tissue, df in val_dict.items():
        st = VAL_TISSUE_SUBTYPES.get(val_tissue)
        if st:
            subtype_dfs.setdefault(st, []).append(df[['Gene', 'Score_Norm']])
            
    # Calculate the mean normalized score for each subtype
    for st, dfs in subtype_dfs.items():
        if dfs:
            combined_df = pd.concat(dfs)
            mean_df = combined_df.groupby('Gene')['Score_Norm'].mean().reset_index()
            subtype_mean_scores[model_family][st] = mean_df

            
            subtype_mean_scores[model_family][st] = mean_df
# ==========================================
#  Core vs. Adaptive Stacked Bar Chart
# ==========================================
adaptation_records = []

# Group raw_attributions by model and subtype so we can extract gene lists
# Structure: {model_family: {subtype: [list_of_genes1, list_of_genes2]}}
subtype_gene_lists = {}
for model_key, val_dict in raw_attributions.items():
    model_family = model_key.split(' ', 1)[0]
    if model_family not in subtype_gene_lists:
        subtype_gene_lists[model_family] = {}
        
    for val_tissue, df in val_dict.items():
        st = VAL_TISSUE_SUBTYPES.get(val_tissue)
        if st:
            # Assumes df is already sorted by importance
            subtype_gene_lists[model_family].setdefault(st, []).append(df['Gene'].tolist())

for model_family, subtypes in subtype_gene_lists.items():
    if baseline_subtype not in subtypes:
        continue
        
    # Get Top 100 for the baseline using rank consensus
    base_consensus = compute_consensus(subtypes[baseline_subtype])
    base_top100 = set(base_consensus.head(100)['Gene'])
    
    for rare_st in rare_subtypes:
        if rare_st not in subtypes:
            continue
            
        # Get Top 100 for the rare subtype using rank consensus
        rare_consensus = compute_consensus(subtypes[rare_st])
        rare_top100 = set(rare_consensus.head(100)['Gene'])
        
        # Calculate overlap and differences
        conserved = len(base_top100.intersection(rare_top100))
        acquired = len(rare_top100.difference(base_top100))
        
        adaptation_records.append({
            'Model': model_family,
            'Comparison': rare_st, 
            'Conserved': conserved,
            'Acquired': acquired
        })

df_adapt = pd.DataFrame(adaptation_records)

# Plotting the Stacked Bar Chart
fig, axes = plt.subplots(1, len(rare_subtypes), figsize=(12, 5), sharey=True)
if len(rare_subtypes) == 1: axes = [axes]

for ax, rare_st in zip(axes, rare_subtypes):
    subset = df_adapt[df_adapt['Comparison'] == f"{baseline_subtype} vs {rare_st}"]
    
    # Bottom bar (Conserved)
    ax.bar(subset['Model'], subset['Conserved'], color='lightgrey', edgecolor='black', label='Conserved (Core)')
    # Top bar (Acquired)
    ax.bar(subset['Model'], subset['Acquired'], bottom=subset['Conserved'], 
           color=[model_color_map.get(m, 'blue') for m in subset['Model']], 
           edgecolor='black', alpha=0.8, label='Acquired (Adaptive)')
    
    # Add value labels
    for i, (cons, acq) in enumerate(zip(subset['Conserved'], subset['Acquired'])):
        ax.text(i, cons / 2, str(cons), ha='center', va='center', fontweight='bold')
        ax.text(i, cons + (acq / 2), str(acq), ha='center', va='center', fontweight='bold', color='white')

    ax.set_title(f"Adaptation: {baseline_subtype} $\\rightarrow$ {rare_st}\n(Top 100 Genes)", pad=15)
    ax.set_ylim(0, 100)
    if ax == axes[0]: ax.set_ylabel("Number of Genes")

# Custom Legend
handles, labels = axes[0].get_legend_handles_labels()
# Just take the first two to avoid duplicates
fig.legend(handles[:2], labels[:2], loc='upper center', bbox_to_anchor=(0.5, 1.05), ncol=2)

plt.tight_layout()
# adapt_path = output_dir / "adaptation_stacked_bars.png"
# fig.savefig(adapt_path, dpi=300, bbox_inches="tight")
plt.close()
# print(f"Stacked bars saved to: {adapt_path}")


# ==========================================
#  Rank-Shift Scatter Plots
# ==========================================
scatter_dir = output_dir / "scatter_plots"
scatter_dir.mkdir(exist_ok=True)

for model_family, subtypes in subtype_mean_scores.items():
    if baseline_subtype not in subtypes:
        continue
        
    df_base = subtypes[baseline_subtype].rename(columns={'Score_Norm': 'Score_Base'})
    
    for rare_st in rare_subtypes:
        if rare_st not in subtypes:
            continue
            
        df_rare = subtypes[rare_st].rename(columns={'Score_Norm': 'Score_Rare'})
        
        # Merge the two dataframes on Gene, filling missing values with 0
        merged = pd.merge(df_base, df_rare, on='Gene', how='outer').fillna(0)
        # --- Min-max normalise each axis independently to [0, 1] ---
        
        merged['Diff'] = merged['Score_Rare'] - merged['Score_Base']
        merged = merged[merged['Diff'] != 0]

        # Normalise both columns together so the global max = 1.0
        global_min = min(merged['Score_Base'].min(), merged['Score_Rare'].min())
        global_max = max(merged['Score_Base'].max(), merged['Score_Rare'].max())

        merged['Score_Base'] = (merged['Score_Base'] - global_min) / (global_max - global_min)
        merged['Score_Rare'] = (merged['Score_Rare'] - global_min) / (global_max - global_min)

        # Recalculate Diff after normalisation
        merged['Diff'] = merged['Score_Rare'] - merged['Score_Base']

        # Exclude genes with no difference between subtypes
        merged = merged[merged['Diff'] != 0]
        # --- Export the data to CSV ---
        csv_safe_name = f"scatter_data_{model_family}_{rare_st}".replace("+", "plus")
        csv_path = scatter_dir / f"{csv_safe_name}.csv"
        # Sort so acquired genes are at the top, dropped at the bottom
        merged_sorted = merged.sort_values('Score_Rare', ascending=False)
        merged_sorted.to_csv(csv_path, index=False)
        # ----------------------------------------
        
        # Exclude Ensembl IDs (start with ENSG) before selecting top 5
        merged_named = merged[~merged['Gene'].str.startswith('ENSG')]

        # Get top 5 acquired and top 5 dropped genes for labelling
        top_acquired = merged_named.sort_values('Diff', ascending=False).head(5)
        top_dropped = merged_named.sort_values('Diff', ascending=True).head(5)
        genes_to_label = pd.concat([top_acquired, top_dropped])
        
        fig, ax = plt.subplots(figsize=(7, 6))
        
        # Plot all genes as faint background dots
        ax.scatter(merged['Score_Base'], merged['Score_Rare'], alpha=0.2, color='grey', s=15)
        
        # Highlight and label the most shifted genes
        ax.scatter(genes_to_label['Score_Base'], genes_to_label['Score_Rare'], 
                   color=model_color_map.get(model_family, 'red'), s=40, edgecolor='black')
      
        texts = []
        for _, row in genes_to_label.iterrows():
            texts.append(ax.text(row['Score_Base'], row['Score_Rare'], row['Gene'],
                                fontsize=9, fontweight='bold'))

        # adjust_text(texts, ax=ax,
        #             arrowprops=dict(arrowstyle='-', color='black', lw=0.8))
        
        adjust_text(texts, ax=ax,
                    arrowprops=dict(arrowstyle='-', color='black', lw=0.8),
                    expand=(1.1, 1.1),      # how much breathing room around each label
                    force_text=(0.5, 0.5),  # how hard to push text away from other text
                    force_points=(0.3, 0.3) # how hard to push text away from points
                    )
            
        # Draw the $y=x$ diagonal line
        ax.plot([0, 1], [0, 1], transform=ax.transAxes, ls='--', color='black', alpha=0.5, zorder=0)
        
        ax.set_title(f"{model_family} Gene Shifts\n{baseline_subtype} vs {rare_st}")
        ax.set_xlabel(f"Mean Score in {baseline_subtype} (Baseline)")
        ax.set_ylabel(f"Mean Score in {rare_st} (Rare Subtype)")
        ax.set_xlim(-0.05, 1.05)
        ax.set_ylim(-0.05, 1.05)
        
        plt.tight_layout()
        safe_name = f"scatter_{model_family}_{rare_st}".replace("+", "plus")
        fig.savefig(scatter_dir / f"{safe_name}.svg", bbox_inches="tight")
        plt.close()

print(f"Scatter plots saved to: {scatter_dir}")
```

```python
# ==========================================
# Idea 1: Core vs. Adaptive Grouped Stacked Bars (All in One Plot)
# ==========================================
import matplotlib.patches as mpatches
import numpy as np
adaptation_configs = {
    "ccRCC": ["Metastatic", "ChRCC"],
    "LUAD": ["LUSC", "SCLC"],
    "ER+": ["HER2+", "TNBC"]
}
models_to_plot = ["scGPT", "Geneformer", "LightGBM"] # Ensures a consistent left-to-right order
bar_width = 0.25
num_configs = len(adaptation_configs)

# Create one wide figure containing subplots for all training conditions
fig, axes = plt.subplots(1, num_configs, figsize=(5 * num_configs, 6), sharey=True)

# Ensure axes is iterable even if you later drop down to 1 config
if num_configs == 1:
    axes = [axes]

for ax, (baseline_subtype, rare_subtypes) in zip(axes, adaptation_configs.items()):
    adaptation_records = []
    
    # 1. Aggregate the data for this specific subplot
    for model_family, subtypes in subtype_mean_scores.items():
        if baseline_subtype not in subtypes:
            continue
            
        base_top100 = set(subtypes[baseline_subtype].sort_values('Score_Norm', ascending=False).head(100)['Gene'])
        
        for rare_st in rare_subtypes:
            if rare_st not in subtypes:
                continue
                
            rare_top100 = set(subtypes[rare_st].sort_values('Score_Norm', ascending=False).head(100)['Gene'])
            
            conserved = len(base_top100.intersection(rare_top100))
            acquired = len(rare_top100.difference(base_top100))
            
            adaptation_records.append({
                'Model': model_family,
                'Rare_Subtype': rare_st,
                'Conserved': conserved,
                'Acquired': acquired
            })

    df_adapt = pd.DataFrame(adaptation_records)
    
    if df_adapt.empty:
        ax.set_title(f"No Data for {baseline_subtype}")
        continue

    # 2. Draw the grouped bars for this subplot
    n_rare = len(rare_subtypes)
    n_models = len(models_to_plot)

    for i, rare_st in enumerate(rare_subtypes):
        subset = df_adapt[df_adapt['Rare_Subtype'] == rare_st]
        
        for j, model in enumerate(models_to_plot):
            row = subset[subset['Model'] == model]
            if row.empty:
                continue
                
            cons = row['Conserved'].values[0]
            acq = row['Acquired'].values[0]
            
            # Calculate offset to group bars side-by-side around the tick mark
            offset = (j - (n_models - 1) / 2) * bar_width
            x_pos = i + offset
            
            # Bottom bar (Conserved)
            ax.bar(x_pos, cons, width=bar_width, color='lightgrey', edgecolor='black')
            
            # Top bar (Acquired)
            model_color = model_color_map.get(model, '#333333')
            ax.bar(x_pos, acq, bottom=cons, width=bar_width, 
                   color=model_color, edgecolor='black', alpha=0.8)
            
            # Add text labels inside the bars
            ax.text(x_pos, cons / 2, str(cons), ha='center', va='center', fontweight='bold', fontsize=9)
            ax.text(x_pos, cons + (acq / 2), str(acq), ha='center', va='center', fontweight='bold', color='white', fontsize=9)

    # 3. Formatting this specific subplot
    ax.set_xticks(np.arange(n_rare))
    # Using an arrow to show the transition (e.g., "-> HER2+")
    ax.set_xticklabels([f"$\\rightarrow$ {st}" for st in rare_subtypes], fontsize=12, fontweight='bold')
    ax.set_title(f"{baseline_subtype}", fontsize=14, pad=15)
    
    # Only label the y-axis on the far left plot
    if ax == axes[0]:
        ax.set_ylabel("Number of Genes (Top 100)", fontsize=12)
        ax.set_ylim(0, 105)

# 4. Create a single master legend for the entire figure
legend_elements = [mpatches.Patch(facecolor='lightgrey', edgecolor='black', label='Conserved (Core)')]
for model in models_to_plot:
    legend_elements.append(
        mpatches.Patch(facecolor=model_color_map.get(model, 'blue'), edgecolor='black', alpha=0.8, label=f'Acquired ({model})')
    )

# Place the legend centrally above the subplots
fig.legend(handles=legend_elements, loc='upper center', bbox_to_anchor=(0.5, 1.07), ncol=4, framealpha=0.9, fontsize=11)

plt.tight_layout()
adapt_path = output_dir / "Figure5_adaptation_grouped_bars_all_conditions.svg"
fig.savefig(adapt_path, bbox_inches="tight")
plt.close()

print(f"Master grouped bar chart saved to: {adapt_path}")
```
