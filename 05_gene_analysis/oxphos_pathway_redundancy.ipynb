```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# ==========================================
# 1. Get OXPHOS gene set from MSigDB via gseapy
# ==========================================

import gseapy as gp

file_map = file_map_scgpt_indomain_breast
gene_sets = gp.get_library(name="MSigDB_Hallmark_2020", organism="Human")
OXPHOS_GENES = set(gene_sets["Oxidative Phosphorylation"])
print(f"OXPHOS gene set size: {len(OXPHOS_GENES)}")

# ==========================================
# 2. Load attributions and extract OXPHOS genes
# ==========================================

TOP_K = 50  # look deeper than top-50 to catch pathway genes

records = []

for dataset_name, path in file_map.items():
    df = pd.read_csv(path)
    
    gene_col  = "GeneSymbol" if "GeneSymbol" in df.columns else "Gene"
    score_col = "TotalAttributionSum" if "TotalAttributionSum" in df.columns else "SHAP_Importance"

    
    df = df.rename(columns={gene_col: "Gene", score_col: "Score"})
    df = df[df["Score"] > 0].sort_values("Score", ascending=False).head(50).reset_index(drop=True)
    df["Rank"] = df.index + 1

    # Filter to OXPHOS genes present in attributions
    oxphos_in_df = df[df["Gene"].isin(OXPHOS_GENES)].copy()
    oxphos_in_df["Dataset"] = dataset_name

    records.append(oxphos_in_df[["Dataset", "Gene", "Score", "Rank"]])
    print(f"{dataset_name:30s}: {len(oxphos_in_df)} OXPHOS genes attributed "
          f"(top-ranked: {oxphos_in_df['Gene'].iloc[0] if len(oxphos_in_df) else 'none'}, "
          f"rank {int(oxphos_in_df['Rank'].iloc[0]) if len(oxphos_in_df) else 'n/a'})")

all_oxphos = pd.concat(records, ignore_index=True)

# ==========================================
# 3. Pivot: genes × datasets, value = rank
#    (NaN = gene not attributed in that dataset)
# ==========================================

pivot_rank = all_oxphos.pivot_table(
    index="Gene", columns="Dataset", values="Rank", aggfunc="min"
)

# Also a score pivot for heatmap colouring
pivot_score = all_oxphos.pivot_table(
    index="Gene", columns="Dataset", values="Score", aggfunc="sum"
).fillna(0)

# Sort genes by how consistently they appear (fewest NaN first, then mean rank)
pivot_rank["n_datasets"] = pivot_rank.notna().sum(axis=1)
pivot_rank["mean_rank"]  = pivot_rank.drop(columns=["n_datasets"]).mean(axis=1)
pivot_rank = pivot_rank.sort_values(["n_datasets", "mean_rank"], ascending=[False, True])
pivot_rank = pivot_rank.drop(columns=["n_datasets", "mean_rank"])

print(f"\nTotal unique OXPHOS genes attributed across all datasets: {len(pivot_rank)}")
print(f"\nGenes attributed in ALL datasets:")
print(pivot_rank.dropna().index.tolist())

# Define desired dataset order — controls x-axis in both heatmaps
DATASET_ORDER = [
    "GSE176078 ER+",
    "GSE161529 ER+",
    "PRJNA1140267 ER+",
    "EMTAB8107 Luminal",
    "GSE176078 HER2+",
    "GSE161529 HER2+",
    "GSE176078 TNBC",
    "GSE161529 TNBC",
    "GSE161529 TNBC BRCA1",
]

# Apply after building pivots — just reindex columns
DATASET_ORDER_AVAILABLE = [d for d in DATASET_ORDER if d in pivot_rank.columns]


pivot_rank  = pivot_rank[DATASET_ORDER_AVAILABLE]
pivot_score = pivot_score[DATASET_ORDER_AVAILABLE]

# ==========================================
# rank heatmap
#    Shows WHERE in the ranked list each gene falls
# ==========================================

fig, ax = plt.subplots(figsize=(max(10, len(file_map) * 1.2),
                                 max(8, len(pivot_rank) * 0.3)))

# Only show genes appearing in at least 2 datasets for clarity
pivot_rank_filtered = pivot_rank[pivot_rank.notna().sum(axis=1) >= 2]

sns.heatmap(
    pivot_rank_filtered,
    cmap="YlOrRd_r",
    ax=ax,
    annot=len(pivot_rank_filtered) < 30,   # only annotate if small enough
    fmt=".0f",
    linewidths=0.3, linecolor="white",
    cbar_kws={"label": "Attribution rank (lower = more important)"},
    yticklabels=True, xticklabels=True
)

ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="right", fontsize=8)
ax.set_yticklabels(ax.get_yticklabels(), fontsize=7)
ax.set_title("OXPHOS Gene Attribution Ranks per Dataset\n"
             "Grey = not attributed, Red = top-ranked",
             fontsize=11, fontweight="bold")

plt.tight_layout()
plt.savefig("Figure10a_oxphos_rank_heatmap_scgpt.svg", bbox_inches="tight")
plt.show()

# ==========================================
# 6. Summary stats — quantify switching
# ==========================================

print("\n=== OXPHOS Attribution Summary ===")
print(f"\nGenes attributed in ALL {len(file_map)} datasets (stable core):")
stable = pivot_rank.dropna()
print(stable.index.tolist())

print(f"\nGenes attributed in only 1 dataset (dataset-specific):")
specific = pivot_rank[pivot_rank.notna().sum(axis=1) == 1]
for gene in specific.index:
    ds = specific.loc[gene].dropna().index[0]
    print(f"  {gene:15s} → only in {ds}")

print(f"\nPer-dataset OXPHOS gene count:")
print(all_oxphos.groupby("Dataset")["Gene"].count().sort_values(ascending=False))

print(f"\nJaccard similarity between datasets (OXPHOS gene overlap):")
datasets = list(file_map.keys())
jaccard_matrix = pd.DataFrame(index=datasets, columns=datasets, dtype=float)
gene_sets_per_ds = {
    ds: set(all_oxphos[all_oxphos["Dataset"] == ds]["Gene"])
    for ds in datasets
}
for d1 in datasets:
    for d2 in datasets:
        s1, s2 = gene_sets_per_ds[d1], gene_sets_per_ds[d2]
        inter = len(s1 & s2)
        union = len(s1 | s2)
        jaccard_matrix.loc[d1, d2] = inter / union if union > 0 else 0.0

print(jaccard_matrix.round(2).to_string())
```
